# TCA cycle model
This model is based on the metabolic network presentet in "OpenFLUX" Lake-Ee Quek et al.. 
First the true data(MIDs) are computed with the flux-values given in figure 3 in the article. 
Then simulation data is made by adding noise to the true data. The simulated data is CLR transformed. 

The simulated data and the CLR transformed simulated data are used to estimated fluxes in the metabolic network. 
In the end the results from the estimated flux on simulated data are compared with the estimated flux from the CLR transformed simulated data.

## Simulations

We simulate MIDs by first assuming normal distributions on the absolute abundance (the reads), and then close it to 1 to get the relative abundance. 

```python
rng = np.random.default_rng(seed=1234)
key = jax.random.PRNGKey(1234)
sims = 1000

total_mol = 1e5
total_isos = jax.tree.map(lambda x: x * total_mol, true_data)
cv = 0.01*total_mol
norm_sigma_cv = jax.tree.map(lambda x: cv, total_isos)

key, key_sim = jax.random.split(key)
# We assume the absolute abandance to be independent normal distributions. 
norm_sim = jax.tree.map(
    lambda mean, sigma: mean + sigma * jax.random.normal(key_sim, (sims, mean.shape[0])),
    total_isos,
    norm_sigma_cv,
)
```

```python
jax.tree.map(
    lambda x: jnp.all(x >0),
    norm_sim
)
```

```python
norm_sim_positive = jax.tree.map(
    lambda x: jnp.clip(x, 1e-8, None),
    norm_sim
)
```

```python
# Simulated absolute abandance are closed to 1 to make relative abundance
iso_proportions = jax.tree.map(
    lambda mat: closure(mat),
    norm_sim_positive,
)

# Check that all sum to 1
jax.tree.map(
    lambda x: jnp.all(jnp.isclose(jnp.sum(x, axis=-1), 1.0, rtol=1e-10, atol=1e-10)),
    iso_proportions
)
```

```python
# We calculate the mean of the relative abundances. 
mean_sim = jax.tree.map(
    lambda x: jnp.mean(x, axis=0),  # mean over isotopolouges
    iso_proportions
)
# and the standard deviation
std_sim = jax.tree.map(
    lambda x: jnp.std(x, axis=0),  # standard deviation over isotopolouges
    iso_proportions
)

# and compute a covariance matrix
D_sim = jax.tree.map(
    lambda x: jnp.diag(x), 
    std_sim
)
```

```python
def objective_wssr_fluxes(params: dict, args):
    data_obs, D_list = args
    outputs = TCA_OpenFLux(params["free"], params["fixed"])
    residuals = jax.tree.map(lambda e, o: e - o, outputs, data_obs)
    total = sum(
        r.T @ jnp.linalg.inv(D) @ r
        for r, D in zip(jax.tree.leaves(residuals),
                        jax.tree.leaves(D_list))
    )
    return total
```

```python
def run_profile_likelihood(flux_index, start_params, objective_fun, phi_optimal, data, D, fixed, direction=1, k_steps=50, step_size=1):
    path = []
    chi2_threshold = 3.84
    
    current_fixed_val = float(start_params[flux_index])
    other_idx = 1 - flux_index
    current_other_guess = jnp.array([start_params[other_idx]])
    
    # Track the MLE point itself
    path.append((current_fixed_val, float(phi_optimal)))

    for k in range(k_steps):
        current_fixed_val += direction * step_size
        if current_fixed_val <= 0:
            break   
        
        # We pass only the parameter to optimize to the solver
        def inner_objective(other_params, args):
            val, d, mat_D, vec_v1 = args
            p = jnp.zeros((2,))
            p = p.at[flux_index].set(val)
            p = p.at[other_idx].set(other_params[0])
            return objective_fun(p, (d, mat_D, vec_v1))

        inner_sol = optimistix.minimise(
            inner_objective,
            solver=optimistix.LevenbergMarquardt(rtol=1e-12, atol=1e-12),
            y0=current_other_guess,
            args=(current_fixed_val, data, D, v1),
            max_steps=10000,
            throw=False
        )
        
        current_other_guess = inner_sol.value
        
        # Reconstruct full vector to calculate final Phi
        p_final = jnp.zeros((2,))
        p_final = p_final.at[flux_index].set(current_fixed_val)
        p_final = p_final.at[other_idx].set(current_other_guess[0])

        phi_u = objective_fun(p_final, (data, D, v1))
        delta_phi = phi_u - phi_optimal
        path.append((float(current_fixed_val), float(phi_u)))

        if delta_phi > chi2_threshold:
            break
        
            
    return path

def add_profile_to_ax(ax, flux_vals, phi_vals, phi_optimal, mle_val, flux_label):
    ax.plot(flux_vals, phi_vals, 'o-', color='teal', markersize=3, label='Profile')
    ax.axvline(mle_val, color='red', linestyle='--', alpha=0.5, label='MLE')
    
    # threshold_val = phi_optimal + 3.84
    # ax.axhline(threshold_val, color='orange', linestyle=':', label=r'95% CI Thresh')
    
    ax.set_title(f"Profile: {flux_label}")
    ax.set_xlabel(f"Value")
    ax.set_ylabel(r"$\Phi(u)$")
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize='small')
```
