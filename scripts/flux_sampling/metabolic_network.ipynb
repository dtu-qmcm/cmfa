## Import Libraries

```python
import cobra
from cobra import Model, Reaction, Metabolite
```

## Create the Model

```python
# Create a new model
model = Model('metabolic_network')
```

### Define Metabolites

```python
# Define all metabolites
metabolites = {
    'PYR_EX': Metabolite('PYR_EX', compartment='e'),
    'GLU_EX': Metabolite('GLU_EX', compartment='e'),
    'PYR': Metabolite('PYR', compartment='c'),
    'AKG': Metabolite('AKG', compartment='c'),
    'ACCOA': Metabolite('ACCOA', compartment='c'),
    'CO2': Metabolite('CO2', compartment='c'),
    'NADH': Metabolite('NADH', compartment='c'),
    'OAA': Metabolite('OAA', compartment='c'),
    'CIT': Metabolite('CIT', compartment='c'),
    'SUC': Metabolite('SUC', compartment='c'),
    'ATP': Metabolite('ATP', compartment='c'),
    'FADH2': Metabolite('FADH2', compartment='c'),
    'O2': Metabolite('O2', compartment='c'),
    'CO2_EX': Metabolite('CO2_EX', compartment='e'),
    'ATPM': Metabolite('ATPM', compartment='c'),
    'PYR_B': Metabolite('PYR_B', compartment='c'),
    'AKG_B': Metabolite('AKG_B', compartment='c'),
    'OAA_B': Metabolite('OAA_B', compartment='c'),
    'VALX': Metabolite('VALX', compartment='c'),
    'LYSX': Metabolite('LYSX', compartment='c'),
    'ASPX': Metabolite('ASPX', compartment='c'),
    'O2_EX': Metabolite('O2_EX', compartment='e'),
}

# Add metabolites to model
model.add_metabolites(metabolites.values())
```

### Define Reactions

```python
# R1: pyruvate uptake
r1 = Reaction('R1')
r1.name = 'Pyruvate uptake'
r1.lower_bound = 0
r1.upper_bound = 1000
r1.add_metabolites({
    metabolites['PYR_EX']: -1,
    metabolites['PYR']: 1
})

# R2: glutamate take
r2 = Reaction('R2')
r2.name = 'Glutamate uptake'
r2.lower_bound = 0
r2.upper_bound = 1000
r2.add_metabolites({
    metabolites['GLU_EX']: -1,
    metabolites['AKG']: 1
})

# R3: pyruvate dehydrogenase
r3 = Reaction('R3')
r3.name = 'Pyruvate dehydrogenase'
r3.lower_bound = 0
r3.upper_bound = 1000
r3.add_metabolites({
    metabolites['PYR']: -1,
    metabolites['ACCOA']: 1,
    metabolites['CO2']: 1,
    metabolites['NADH']: 1
})

# R4: citrate synthase
r4 = Reaction('R4')
r4.name = 'Citrate synthase'
r4.lower_bound = 0
r4.upper_bound = 1000
r4.add_metabolites({
    metabolites['ACCOA']: -1,
    metabolites['OAA']: -1,
    metabolites['CIT']: 1
})

# R5: iso-citrate dehydrogase
r5 = Reaction('R5')
r5.name = 'Isocitrate dehydrogenase'
r5.lower_bound = 0
r5.upper_bound = 1000
r5.add_metabolites({
    metabolites['CIT']: -1,
    metabolites['AKG']: 1,
    metabolites['CO2']: 1,
    metabolites['NADH']: 1
})

# R6: alpha-ketoglutarate dehydrogenase & succinyl-CoA hydrolase
r6 = Reaction('R6')
r6.name = 'Alpha-ketoglutarate dehydrogenase & succinyl-CoA hydrolase'
r6.lower_bound = 0
r6.upper_bound = 1000
r6.add_metabolites({
    metabolites['AKG']: -1,
    metabolites['SUC']: 1,
    metabolites['CO2']: 1,
    metabolites['NADH']: 1,
    metabolites['ATP']: 1
})

# R7: fumarate hydratase & malate dehydrogenase
r7 = Reaction('R7')
r7.name = 'Fumarate hydratase & malate dehydrogenase'
r7.lower_bound = 0
r7.upper_bound = 1000
r7.add_metabolites({
    metabolites['SUC']: -1,
    metabolites['OAA']: 1,
    metabolites['FADH2']: 1,
    metabolites['NADH']: 1
})

# R8: fumarate hydratase & malate dehydrogenase (reverse)
r8 = Reaction('R8')
r8.name = 'Fumarate hydratase & malate dehydrogenase (reverse)'
r8.lower_bound = 0
r8.upper_bound = 1000
r8.add_metabolites({
    metabolites['OAA']: -1,
    metabolites['FADH2']: -1,
    metabolites['NADH']: -1,
    metabolites['SUC']: 1
})

# R9: malic enzyme
r9 = Reaction('R9')
r9.name = 'Malic enzyme'
r9.lower_bound = 0
r9.upper_bound = 1000
r9.add_metabolites({
    metabolites['OAA']: -1,
    metabolites['PYR']: 1,
    metabolites['CO2']: 1
})

# R10: pyruvate carboxylase
r10 = Reaction('R10')
r10.name = 'Pyruvate carboxylase'
r10.lower_bound = 0
r10.upper_bound = 1000
r10.add_metabolites({
    metabolites['PYR']: -1,
    metabolites['CO2']: -1,
    metabolites['ATP']: -1,
    metabolites['OAA']: 1
})

# R11: oxidative phosphorylation
r11 = Reaction('R11')
r11.name = 'Oxidative phosphorylation'
r11.lower_bound = 0
r11.upper_bound = 1000
r11.add_metabolites({
    metabolites['NADH']: -2,
    metabolites['O2']: -1,
    metabolites['ATP']: 4
})

# R12: oxidative phosphorylation
r12 = Reaction('R12')
r12.name = 'Oxidative phosphorylation'
r12.lower_bound = 0
r12.upper_bound = 1000
r12.add_metabolites({
    metabolites['FADH2']: -2,
    metabolites['O2']: -1,
    metabolites['ATP']: 2
})

# R13: oxygen uptake
r13 = Reaction('R13')
r13.name = 'Oxygen uptake'
r13.lower_bound = 0
r13.upper_bound = 1000
r13.add_metabolites({
    metabolites['O2_EX']: -1,
    metabolites['O2']: 1
})

# R14: CO2 evolution
r14 = Reaction('R14')
r14.name = 'CO2 evolution'
r14.lower_bound = 0
r14.upper_bound = 1000
r14.add_metabolites({
    metabolites['CO2']: -1,
    metabolites['CO2_EX']: 1
})

# R15: ATP maintenance
r15 = Reaction('R15')
r15.name = 'ATP maintenance'
r15.lower_bound = 0
r15.upper_bound = 1000
r15.add_metabolites({
    metabolites['ATP']: -1,
    metabolites['ATPM']: 1
})

# R16: pyruvate biomass drain
r16 = Reaction('R16')
r16.name = 'Pyruvate biomass drain'
r16.lower_bound = 0
r16.upper_bound = 1000
r16.add_metabolites({
    metabolites['PYR']: -1,
    metabolites['PYR_B']: 1
})

# R17: alpha-ketoglutarate biomass drain
r17 = Reaction('R17')
r17.name = 'Alpha-ketoglutarate biomass drain'
r17.lower_bound = 0
r17.upper_bound = 1000
r17.add_metabolites({
    metabolites['AKG']: -1,
    metabolites['AKG_B']: 1
})

# R18: oxaloacetate biomass drain
r18 = Reaction('R18')
r18.name = 'Oxaloacetate biomass drain'
r18.lower_bound = 0
r18.upper_bound = 1000
r18.add_metabolites({
    metabolites['OAA']: -1,
    metabolites['OAA_B']: 1
})

# R19: Valine synthesis
r19 = Reaction('R19')
r19.name = 'Valine synthesis'
r19.lower_bound = 0
r19.upper_bound = 1000
r19.add_metabolites({
    metabolites['PYR']: -2,
    metabolites['VALX']: 1,
    metabolites['CO2']: 1
})

# R20: Lysine synthesis
r20 = Reaction('R20')
r20.name = 'Lysine synthesis'
r20.lower_bound = 0
r20.upper_bound = 1000
r20.add_metabolites({
    metabolites['OAA']: -1,
    metabolites['PYR']: -1,
    metabolites['LYSX']: 1,
    metabolites['CO2']: 1
})

# R21: Aspartate synthesis
r21 = Reaction('R21')
r21.name = 'Aspartate synthesis'
r21.lower_bound = 0
r21.upper_bound = 1000
r21.add_metabolites({
    metabolites['OAA']: -1,
    metabolites['ASPX']: 1
})

# Add all reactions to model
reactions = [r1, r2, r3, r4, r5, r6, r7, r8, r9, r10, r11, r12, r13, r14, r15, r16, r17, r18, r19, r20, r21]
model.add_reactions(reactions)
```

## Set Objective Function

```python
# Set objective to maximize ATPM (ATP maintenance/biomass production)
model.objective = 'R15'
print(f"Objective: {model.objective}")
```

## Model Summary

```python
# Print model summary
print("=== Model Summary ===")
print(f"Model ID: {model.id}")
print(f"Reactions: {len(model.reactions)}")
print(f"Metabolites: {len(model.metabolites)}")
print(f"Genes: {len(model.genes)}")
print()

print("=== Reactions ===")
for rxn in model.reactions:
    print(f"{rxn.id}: {rxn.reaction}")
```

## Test FBA Simulation

```python
# Test 1: Basic FBA with PYR_EX uptake
print("=== Test 1: PYR uptake only ===")
with model:
    model.reactions.R1.lower_bound = 10  # Set PYR_EX uptake
    model.reactions.R2.lower_bound = 0   # No GLU_EX uptake
    model.reactions.R13.lower_bound = 100  # Unlimited O2
    solution = model.optimize()
    print(f"Objective value (ATPM): {solution.objective_value:.4f}")
    print(f"Status: {solution.status}")
    if solution.status == 'optimal':
        print("\nTop fluxes:")
        for rxn_id, flux in solution.fluxes.abs().sort_values(ascending=False).head(10).items():
            if flux > 0.001:
                print(f"  {rxn_id}: {solution.fluxes[rxn_id]:.4f}")
```

```python
# Test 2: GLU uptake only
print("=== Test 2: GLU uptake only ===")
with model:
    model.reactions.R1.lower_bound = 0   # No PYR_EX uptake
    model.reactions.R2.lower_bound = 10  # Set GLU_EX uptake
    model.reactions.R13.lower_bound = 100  # Unlimited O2
    solution = model.optimize()
    print(f"Objective value (ATPM): {solution.objective_value:.4f}")
    print(f"Status: {solution.status}")
    if solution.status == 'optimal':
        print("\nTop fluxes:")
        for rxn_id, flux in solution.fluxes.abs().sort_values(ascending=False).head(10).items():
            if flux > 0.001:
                print(f"  {rxn_id}: {solution.fluxes[rxn_id]:.4f}")
```

```python
# Test 3: Both substrates
print("=== Test 3: Both PYR and GLU uptake ===")
with model:
    model.reactions.R1.lower_bound = 10  # PYR_EX uptake
    model.reactions.R2.lower_bound = 10  # GLU_EX uptake
    model.reactions.R13.lower_bound = 100  # Unlimited O2
    solution = model.optimize()
    print(f"Objective value (ATPM): {solution.objective_value:.4f}")
    print(f"Status: {solution.status}")
    if solution.status == 'optimal':
        print("\nTop fluxes:")
        for rxn_id, flux in solution.fluxes.abs().sort_values(ascending=False).head(10).items():
            if flux > 0.001:
                print(f"  {rxn_id}: {solution.fluxes[rxn_id]:.4f}")
```

```python
# Test 4: Product formation - Valine
print("=== Test 4: Valine production with PYR ===")
with model:
    model.reactions.R1.lower_bound = 10
    model.reactions.R13.lower_bound = 100
    model.reactions.R15.lower_bound = 5  # ATP maintenance requirement
    model.objective = 'R19'  # Maximize VALX production
    solution = model.optimize()
    print(f"Objective value (VALX): {solution.objective_value:.4f}")
    print(f"Status: {solution.status}")
    if solution.status == 'optimal':
        print("\nFlux distribution:")
        for rxn_id in ['R1', 'R3', 'R19', 'R11', 'R15']:
            print(f"  {rxn_id}: {solution.fluxes[rxn_id]:.4f}")
```

```python
# Test 5: Product formation - Lysine
print("=== Test 5: Lysine production ===")
with model:
    model.reactions.R1.lower_bound = 10
    model.reactions.R13.lower_bound = 100
    model.reactions.R15.lower_bound = 5
    model.objective = 'R20'  # Maximize LYSX production
    solution = model.optimize()
    print(f"Objective value (LYSX): {solution.objective_value:.4f}")
    print(f"Status: {solution.status}")
    if solution.status == 'optimal':
        print("\nFlux distribution:")
        for rxn_id in ['R1', 'R3', 'R4', 'R20', 'R11', 'R15']:
            print(f"  {rxn_id}: {solution.fluxes[rxn_id]:.4f}")
```
